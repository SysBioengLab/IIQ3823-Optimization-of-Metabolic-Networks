```python
# Importar librerías de gurobi
import gurobipy as gp
from gurobipy import GRB
```

## Resolver el problema LP

Minimizar:
$$ x - 4y_1 - 3y_2 - 4y_3 $$

Sujeto a:
\begin{align*}
x + y_1 + y_2 + y_3 &\geq 2 \\
x + 6y_1 + 4y_2 +5y_3 &\leq 13 \\
\end{align*}
$$ x \geq 0,\quad y_1,y_2,y_3 \in \{0,1\}$$

### a) Resolver el problema MILP

```python
# Crear el modelo
m1 = gp.Model("Clase6_MILP")

# Variables
x  = m1.addVar(lb=0, vtype=GRB.CONTINUOUS, name="x")
y1 = m1.addVar(vtype=GRB.BINARY, name="y1")
y2 = m1.addVar(vtype=GRB.BINARY, name="y2")
y3 = m1.addVar(vtype=GRB.BINARY, name="y3")
m1.update()

# Restricciones
m1.addConstr(x + y1 + y2 + y3 >= 2, name="c1")
m1.addConstr(x + 6*y1 + 4*y2 + 5*y3 <= 13, name="c2")

# Función objetivo
m1.setObjective(x - 4*y1 - 3*y2 - 4*y3, GRB.MINIMIZE)
```

```python
# Optimizar
# m1.setParam('OutputFlag', 0) # Opcional
m1.optimize()
```

```python
# Mostrar resultados
print(f"Valor óptimo: {m1.ObjVal:.2f}")
print(f"x = {x.X:.2f}")
print(f"y1 = {y1.X:.0f}")
print(f"y2 = {y2.X:.0f}")
print(f"y3 = {y3.X:.0f}")
```

### b) Resolver problema LP relajado

```python
# Definir problema LP relajado
m2 = gp.Model("Clase6_MILP_Relajado")

# Variables
x_relax  = m2.addVar(lb=0, vtype=GRB.CONTINUOUS, name="x")
y1_relax = m2.addVar(lb=0, ub=1, vtype=GRB.CONTINUOUS, name="y1")
y2_relax = m2.addVar(lb=0, ub=1, vtype=GRB.CONTINUOUS, name="y2")
y3_relax = m2.addVar(lb=0, ub=1, vtype=GRB.CONTINUOUS, name="y3")
m2.update()

# Restricciones
m2.addConstr(x_relax + y1_relax + y2_relax + y3_relax >= 2, name="c1")
m2.addConstr(x_relax + 6*y1_relax + 4*y2_relax + 5*y3_relax <= 13, name="c2")

# Función objetivo
m2.setObjective(x_relax - 4*y1_relax - 3*y2_relax - 4*y3_relax, GRB.MINIMIZE)
```

```python
# Optimizar
# m2.setParam('OutputFlag', 0) # Opcional
m2.optimize()
```

```python
# Mostrar resultados
print(f"Valor óptimo: {m2.ObjVal:.2f}")
print(f"x = {x_relax.X:.2f}")
print(f"y1 = {y1_relax.X:.3f}")
print(f"y2 = {y2_relax.X:.3f}")
print(f"y3 = {y3_relax.X:.3f}")
```

### c) Resolver problema MILP sin variables en la función objetivo

```python
# Crear el modelo
m3 = gp.Model("Clase6_MILP_final")

# Variables
x_final  = m3.addVar(lb=0, vtype=GRB.CONTINUOUS, name="x")
y1_final = m3.addVar(vtype=GRB.BINARY, name="y1")
y2_final = m3.addVar(vtype=GRB.BINARY, name="y2")
y3_final = m3.addVar(vtype=GRB.BINARY, name="y3")
m3.update()

# Restricciones
m3.addConstr(x_final + y1_final + y2_final + y3_final >= 2, name="c1")
m3.addConstr(x_final + 6*y1_final + 4*y2_final + 5*y3_final <= 13, name="c2")

# Función objetivo
m3.setObjective(x_final, GRB.MINIMIZE)
```

```python
# Optimizar
# m3.setParam('OutputFlag', 0) # Opcional
m3.optimize()
```

```python
# Mostrar resultados
print(f"Valor óptimo: {m3.ObjVal:.2f}")
print(f"x = {x_final.X:.2f}")
print(f"y1 = {y1_final.X:.0f}")
print(f"y2 = {y2_final.X:.0f}")
print(f"y3 = {y3_final.X:.0f}")
```
