```python
# Importar librerías de gurobi
import gurobipy as gp
from gurobipy import GRB
```

## Resolver problema Primal (Clase 4)

**Problema Primal**

Maximizar:
$$  4x_1 + 3x_2 $$

Sujeto a:
\begin{align*}
x_1 + 2x_2 &\leq 7 \\
-2x_1 - x_2 &\leq 5 \\
5x_1 + 2x_2 &\leq 16 \\
\end{align*}
$$ x_1 \geq 0,\quad x_2 \geq 0$$


```python
# Crear el modelo
primal = gp.Model("LP_clase4")

# Crear variables
x1 = primal.addVar(lb=0.0, ub = 1e6, name="x1") # Límite superior muy grande
x2 = primal.addVar(lb=0.0, ub = 1e6, name="x2")
primal.update()

# Crear restricciones
restriccion1 = primal.addConstr(   x1 + 2*x2 <= 7,  name = "c1")
restriccion2 = primal.addConstr(-2*x1 -   x2 <= 5,  name = "c2")
restriccion3 = primal.addConstr( 5*x1 + 2*x2 <= 16, name = "c3")

# Crear función objetivo
primal.setObjective(4*x1 + 3*x2, GRB.MAXIMIZE)

# Optimizar modelo
primal.optimize()
```

```python
# Imprimir resultados
print(f"\nObj: {primal.ObjVal}")
print(f"x1: {x1.X}")
print(f"x2: {x2.X}")
```

## Resolver problema Dual (Clase 5)

El problema dual es el siguiente:

Minimizar:

$$7y_1 + 5y_2 + 16y_3$$

Sujeto a:

\begin{align*}
y_1 - 2y_2 + 5y_3 &\geq 4 \\
2y_1 - y_2 + 2y_3 &\geq 3 \\
\end{align*}
$$y_1 \geq 0,\quad y_2 \geq 0,\quad y_3 \geq 0 $$


```python
# Crear el modelo dual
dual = gp.Model("Dual_clase5")

# Crear variables duales (una por cada restricción del primal)
y1 = dual.addVar(lb=0.0, ub = 1e6, name="y1")
y2 = dual.addVar(lb=0.0, ub = 1e6, name="y2")
y3 = dual.addVar(lb=0.0, ub = 1e6, name="y3")
dual.update()

# Restricciones duales (una por cada variable primal)
dual.addConstr(y1 - 2*y2 + 5*y3 >= 4, name="dual_c1")
dual.addConstr(2*y1 - y2 + 2*y3 >= 3, name="dual_c2")

# Función objetivo (minimizar porque el primal maximiza)
dual.setObjective(7*y1 + 5*y2 + 16*y3, GRB.MINIMIZE)

# Optimizar modelo dual
dual.optimize()
```

```python
# Imprimir resultados duales
print(f"\nObj dual: {dual.ObjVal:.4f}")
print(f"y1: {y1.X:.4f}")
print(f"y2: {y2.X:.4f}")
print(f"y3: {y3.X:.4f}")
```

## Comparar soluciones

```python
print("Solución Primal:")
print(f"  Objetivo Primal: {primal.ObjVal:.4f}")
print(f"  x1: {x1.X:.4f}")
print(f"  x2: {x2.X:.4f}")

print("\nSolución Dual:")
print(f"  Objetivo Dual: {dual.ObjVal:.4f}")
print(f"  y1: {y1.X:.4f}")
print(f"  y2: {y2.X:.4f}")
print(f"  y3: {y3.X:.4f}")
# Notar que los valores óptimos del problema primal y dual son iguales, 
# lo que confirma la dualidad fuerte.

print("\nCondiciones KKT (complementariedad):")
print(f"  Restricción 1: slack primal = {restriccion1.Slack:.4f}, y1 (variable dual) = {y1.X:.4f}, producto = {restriccion1.Slack * y1.X:.4f}")
print(f"  Restricción 2: slack primal = {restriccion2.Slack:.4f}, y2 (variable dual) = {y2.X:.4f}, producto = {restriccion2.Slack * y2.X:.4f}")
print(f"  Restricción 3: slack primal = {restriccion3.Slack:.4f}, y3 (variable dual) = {y3.X:.4f}, producto = {restriccion3.Slack * y3.X:.4f}")
# Se verifica el teorema de complementariedad de holguras (todos los productos son cero)
```

## Extra

Gurobi trae incluido funciones para extraer las holguras y los valores pi (variables duales) del modelo primal. Para ello, pueden acceder a los atributos de las restricciones usando `Slack` y `Pi`:

```python
for c in primal.getConstrs():
    print(f"{c.ConstrName}: dual = {c.Pi}")

for c in primal.getConstrs():
    print(f"{c.ConstrName}: slack = {c.Slack}")
```

Por último, gurobi tiene una función para transformar un problema primal a dual con `.dualize()`. Sin embargo, no se recomienda tanto su uso porque tiene problemas aún con nombres e identificación de variables. De todos modos, les dejo el código por si quieren ver cómo se hace:

```python
dual_using_gurobi = primal.dualize()

# Para ver las expresiones, se tiene que imprimir en un archivo, porque los métodos de display de 
# Gurobi arrojan error al acceder al nombre de las variables y restricciones del modelo. 
dual_using_gurobi.write("dual.lp")

# Notar que como añadimos un upper bound a las variables, estas apareceran en el modelo dual
# a diferencia de lo que hicimos manualmente.
```

```python
dual_using_gurobi.optimize()
```
